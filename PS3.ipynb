```python
import requests
import pandas as pd
```

```python
# API call
api_url_le = "http://api.worldbank.org/v2/country/USA;WLD/indicator/SP.DYN.LE00.IN?format=json&date=1990:2023"
response_le = requests.get(api_url_le)
response_le.raise_for_status()
data_le = response_le.json()[1]

# Extract relevant data
life_expectancy_data = [
    {"Country": item['country']['value'], "Year": int(item['date']), "Life Expectancy": item['value']}
    for item in data_le if item['value'] is not None
]

# Convert to DataFrame
life_df = pd.DataFrame(life_expectancy_data)

# Pivot the DataFrame
life_df_pivot = life_df.pivot(index="Year", columns="Country", values="Life Expectancy").reset_index()

# Rename columns
life_df_pivot.columns = ["Year", "Life Expectancy for United States", "Life Expectancy for World"]

# Add 'Country' and 'World' columns
life_df_pivot["Country"] = "United States"
life_df_pivot["World"] = "World"

# Reorder columns
life_df_pivot = life_df_pivot[["Year", "Country", "World", "Life Expectancy for United States", "Life Expectancy for World"]]

# Display the DataFrame
life_df_pivot.head()
```

```python
import requests
import pandas as pd

# Dataset ID for Historical Precipitation in Puget Sound Lowlands
dataset_id = "06c44f9a-aae7-401e-874c-de13b7764959"

# API endpoint to query data
url = f"https://api.resourcewatch.org/v1/query/{dataset_id}?sql=SELECT * FROM {dataset_id}"

# Send GET request
response = requests.get(url)

# Check if request is successful
if response.status_code == 200:
    data = response.json()
    rows = data.get("data", [])

    # Convert to DataFrame
    df = pd.DataFrame(rows)

    # Filter data for years between 1990 and 2023
    df_filtered = df[(df['year'] >= 1990) & (df['year'] <= 2023)]

    # Print first few rows of the filtered data
    print(df_filtered)
else:
    print(f"Error: {response.status_code}, {response.text}")
```
