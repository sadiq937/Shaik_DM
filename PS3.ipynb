### Research Overview: The Impact of Coal Consumption, Emissions, and Precipitation on Life Expectancy in the U.S. (1990-2023)

#### Research Motivation & Dataset Selection  
This study explores the interplay between coal consumption, carbon emissions, precipitation, and life expectancy in the U.S. from 1990 to 2023. The selection of these datasets is based on understanding the complex relationship between environmental factors and public health. The datasets used are:

- **Life Expectancy** data from the World Bank, providing insight into long-term health outcomes.
- **Carbon Emissions** data from the U.S., reflecting environmental pollution and its possible health impacts.
- **Coal Consumption** data, indicating the level of coal use, which is a major contributor to both emissions and health risks.
- **Precipitation** data, which may affect public health through water quality, agricultural productivity, and extreme weather events.

#### Research Questions & Hypotheses  
The study addresses several research questions:

1. **How does coal consumption impact life expectancy in the U.S.?**
   - *Hypothesis*: Coal consumption negatively affects life expectancy due to air pollution and health risks.
  
2. **What is the relationship between carbon emissions and life expectancy?**
   - *Hypothesis*: Higher carbon emissions correlate with lower life expectancy, due to the harmful effects of poor air quality.

3. **How does precipitation influence life expectancy in the U.S.?**
   - *Hypothesis*: Changes in precipitation patterns could affect life expectancy indirectly through water and food safety and weather-related events.

4. **How do coal consumption, carbon emissions, and precipitation interact to affect life expectancy?**
   - *Hypothesis*: The combined effects of coal consumption, emissions, and precipitation can influence public health outcomes significantly.

#### Data Merging Methodology  
To investigate these questions, we performed several merges to combine the datasets:

1. **Merge on Life Expectancy and Carbon Emissions**  
   The first merge was performed on the "Year" column to combine the life expectancy and carbon emissions datasets. This provided a dataset where life expectancy data is aligned with emissions for each year.

2. **Merge on Life Expectancy, Carbon Emissions, and Coal Consumption**  
   A second merge integrated coal consumption data with the life expectancy and carbon emissions data. This allowed for the analysis of the effects of coal consumption on both emissions and life expectancy.

3. **Many-to-Many Merge on Life Expectancy, Carbon Emissions, and Precipitation**  
   A **many-to-many** merge was performed between the life expectancy, carbon emissions, and precipitation datasets on the "Year" column. This merge captures the interaction of these environmental factors with life expectancy over multiple years, allowing for more nuanced analysis. The result is a comprehensive dataset that combines all three factors, providing insights into their combined impact on public health.

#### Key Variables  
- **Life Expectancy**: A measure of public health outcomes over time.
- **Emissions in United States**: Reflects the environmental impact, measured as carbon emissions.
- **Coal Consumption**: Represents the level of coal use, contributing to both emissions and health risks.
- **Precipitation**: Provides information on environmental conditions that could affect public health.

#### Conclusion  
This study uses merged datasets to analyze the relationships between coal consumption, carbon emissions, precipitation, and life expectancy in the U.S. The data-driven insights gained from these merges will help understand how environmental factors influence public health and guide policy recommendations for addressing climate change and its health impacts.

## 1. **Loading DataSets**

#### **1. Life Expectancy(Year) Dataset**

About Dataset:

*  Origin: Data sourced from the World Bank API (https://databank.worldbank.org/reports.aspx?source=2&series=SP.DYN.LE00.IN&country=)
*  Content: Contains life expectancy at birth (years).
*  Scope: Includes data for both the United States and the World.
*  Timeframe: Covers the years 1990 to 2023.
*  Structure: Organized with "Country," "Year," and "Life Expectancy" columns

```python
api_url_le = "http://api.worldbank.org/v2/country/USA;WLD/indicator/SP.DYN.LE00.IN?format=json&date=1990:2023"

try:
    response_le = requests.get(api_url_le)
    response_le.raise_for_status()  # Raise HTTPError for bad responses (4xx or 5xx)
    data_le = response_le.json()[1]  # Extract the relevant data

    life_expectancy_data = []
    for item in data_le:
        if item['value'] is not None:
            life_expectancy_data.append({
                "Country": item['country']['value'],
                "Year": item['date'],
                "Life Expectancy": item['value']
            })

    life_df = pd.DataFrame(life_expectancy_data)
    life_df['Year'] = pd.to_numeric(life_df['Year'])

except requests.exceptions.RequestException as e:
    f"Error during API request: {e}"
except (KeyError, IndexError) as e:
    f"Error parsing API response: {e}"
except Exception as e:
    f"An unexpected error occurred: {e}"
life_df.head()
```

#### **3. Historical Precipitation in Puget Sound Lowlands Dataset**

About Dataset:

*  Origin: Resource Watch API, dataset ID "06c44f9a-aae7-401e-874c-de13b7764959" (https://resourcewatch.org/data/explore?section=All+data&selectedCollection=&zoom=3&lat=0&lng=0&pitch=0&bearing=0&basemap=dark&labels=light&aoi=&page=1&sort=most-viewed&sortDirection=-1&topics=%255B%2522precipitation%2522%255D)
*  Content: Historical precipitation data specifically for the Puget Sound Lowlands region.
*  Scope: Focused on precipitation measurements within the specified geographic area.
*  Timeframe: Covers the years 1990 to 2023.
*  Structure:  Data retrieved via API and organized into a pandas DataFrame, with columns likely including year and precipitation values.


```python
import requests
import pandas as pd

# Dataset ID for Historical Precipitation in Puget Sound Lowlands
dataset_id = "06c44f9a-aae7-401e-874c-de13b7764959"

# API endpoint to query data
url = f"https://api.resourcewatch.org/v1/query/{dataset_id}?sql=SELECT * FROM {dataset_id}"

# Send GET request
response = requests.get(url)

# Check if request is successful
if response.status_code == 200:
    data = response.json()
    rows = data.get("data", [])

    # Convert to DataFrame
    prec_df = pd.DataFrame(rows)

    # Filter data for years between 1990 and 2023
    prec_df_filtered = prec_df[(prec_df['year'] >= 1990) & (prec_df['year'] <= 2023)]

else:
    f"Error: {response.status_code}, {response.text}"
```

## 1-to Many Merge on Life Expectancy + Carbon Emissions + Coal Consumption

***The Impact of Coal Consumption on Life Expectancy and Emissions in the U.S. (1990-2023)***

```python
# Rename the 'period' column in coal_df to 'Year' for consistency
coal_df.rename(columns={'period': 'Year'}, inplace=True)

# Change the data type of the 'period' column to integer
coal_df['Year'] = coal_df['Year'].astype(int)

# Verify the change by checking the data type
coal_df['Year'].dtype


# Perform a 1:M merge on 'Year'
merged_coal_emissions = pd.merge(first_merged_df, coal_df, on='Year', how='left')

# Inspect the first few rows of the merged dataframe
merged_coal_emissions.head()
```

```python
# Filter the columns to keep only relevant ones
merged_coal_emissions = merged_coal_emissions[["Year", "Life Expectancy", "Emissions in United States", "consumption"]]

# Display the first few rows to verify
merged_coal_emissions.tail()
```

```python
# Annual Coal Consumption, Emissions, and Life Expectancy in the U.S. (1990-2023)

# Group by 'Year' and sum coal consumption across all states for each year
aggregated_coal_consumption = merged_coal_emissions.groupby("Year").agg(
    {"Life Expectancy": "first",  # Keep the first value for Life Expectancy since it will be the same for each year
     "Emissions in United States": "first",  # Similarly, take the first value for emissions
     "consumption": "sum"}  # Sum the coal consumption for each year
).reset_index()

aggregated_coal_consumption.head()
```

```python
# Convert "consumption" to numeric, forcing errors to NaN
aggregated_coal_consumption["consumption"] = pd.to_numeric(aggregated_coal_consumption["consumption"], errors="coerce")

# Drop rows with NaN values in "Life Expectancy", "Emissions in United States", or "consumption"
aggregated_coal_consumption_cleaned = aggregated_coal_consumption.dropna(subset=["Life Expectancy", "Emissions in United States", "consumption"])

# Create a DataFrame for correlation calculation
correlation_df = aggregated_coal_consumption_cleaned[["Life Expectancy", "Emissions in United States", "consumption"]].corr()

# Create a heatmap for correlation
import plotly.figure_factory as ff
fig = ff.create_annotated_heatmap(
    z=correlation_df.values,
    x=correlation_df.columns.tolist(),
    y=correlation_df.columns.tolist(),
    colorscale="Viridis",
    # title="Correlation Heatmap: Life Expectancy, Emissions, and Coal Consumption"  # Remove this line
)
fig.update_layout(title_text="Correlation Heatmap: Life Expectancy, Emissions, and Coal Consumption")  # Add this line to set title
fig.show()
```
